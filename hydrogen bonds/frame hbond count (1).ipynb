```python
import pandas as pd

df = pd.read_csv(r'C:\Users\Lenovo\Desktop\aminoatom\aminoatom_run2.csv')

frame_col = 'Frame'
residue1_col = 'Residue1_Residue'
residue3_col = 'Residue3_Residue'

results_list = []

current_count = 0
start_frame = None
count1=0
count2=0
count=0
frames=0

for i in range(len(df) - 1):
    if (((df[frame_col][i + 1] == df[frame_col][i] + 1) or(df[frame_col][i + 1] == df[frame_col][i])) and
        ((df[residue1_col][i] == df[residue1_col][i + 1] and df[residue3_col][i] == df[residue3_col][i + 1]) or
         (df[residue1_col][i] == df[residue3_col][i + 1] and df[residue3_col][i] == df[residue1_col][i + 1]))):
        count+=1
        if (df.iloc[i]['Frame']== start_frame):
             count1 +=1
        if (df.iloc[i]['Frame'] == df.iloc[i-1]['Frame'] + 1): 
             count2+=1
                #frames+=1
        #current_count += 1
        #count+=count1+count2
    
        if start_frame is None:
            start_frame = df[frame_col][i]
    else:
        if start_frame is not None:  
            total_frame = df[frame_col][i] - start_frame + 1
            if total_frame >= 3:
                results_list.append({
                    'Residue1': df[residue1_col][i],
                    'Residue3': df[residue3_col][i],
                    'hbond count': count+1,
                    'first frame': start_frame,
                    'final frame': df[frame_col][i],
                    'total frame': total_frame
                })
        
        #current_count = 0
        count1=0
        count2=0
        count=0
        start_frame = None

if start_frame is not None:
    total_frame = df[frame_col].iloc[-1] - start_frame + 1
    if total_frame >= 3:
        results_list.append({
            'Residue1': df[residue1_col].iloc[-1],
            'Residue3': df[residue3_col].iloc[-1],
            'hbond count': count+1,
            'first frame': start_frame,
            'final frame': df[frame_col].iloc[-1],
            'total frame': total_frame
        })

results = pd.DataFrame(results_list)
results.to_csv('hbond amino10h2o2_run2.csv', index=False)
```
