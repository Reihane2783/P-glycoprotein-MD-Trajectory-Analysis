```python
import pandas as pd

# بارگذاری فایل CSV موجود
input_file = r'C:\Users\Lenovo\Desktop\amino_run3.csv' # نام فایل ورودی
data = pd.read_csv(input_file)

# نام ستون‌های حاوی آمینو اسیدها
residue_columns = ['1_Residue' ,'3_Residue']  # لیست ستون‌ها

# ایجاد یک DataFrame جدید برای ذخیره نتایج
result_df = pd.DataFrame()

# شمارش تکرارهای هر آمینو اسید در هر ستون
for column in residue_columns:
    # شمارش تعداد تکرارهای آمینو اسیدها در ستون
    counts = data[column].value_counts()
    
    # تبدیل به DataFrame و تغییر نام ستون‌ها
    counts_df = counts.reset_index()
    counts_df.columns = ['Amino Acid', column + ' Count']
    
    # ادغام نتایج با DataFrame اصلی
    if result_df.empty:
        result_df = counts_df
    else:
        result_df = pd.merge(result_df, counts_df, on='Amino Acid', how='outer')

# ذخیره‌سازی به یک فایل CSV جدید
output_file = 'amino_acid_countsrun3.csv'  # نام فایل خروجی
result_df.to_csv(output_file, index=False, encoding='utf-8-sig')

print(f'فایل {output_file} با موفقیت ایجاد شد.')
```
