```python
import pandas as pd

# بارگذاری فایل CSV
df = pd.read_csv(r'C:\Users\Lenovo\Desktop\biophys\5\run2\amino_hbond_5H2O2_run2.csv')

# تعریف ستونها
frame_col = 'Frame'
residue1_col = 'Residue1_Residue'
residue3_col = 'Residue3_Residue'

# ایجاد یک لیست خالی برای ذخیره نتایج
results_list = []

# شناسایی فریمهای متوالی
current_count = 0
start_frame = None

for i in range(len(df) - 1):
    # بررسی اینکه آیا فریم فعلی و فریم بعدی متوالی هستند
    if (((df[frame_col][i + 1] == df[frame_col][i] + 1) or(df[frame_col][i + 1] == df[frame_col][i])) and
        ((df[residue1_col][i] == df[residue1_col][i + 1] and df[residue3_col][i] == df[residue3_col][i + 1]) or
         (df[residue1_col][i] == df[residue3_col][i + 1] and df[residue3_col][i] == df[residue1_col][i + 1]))):
        
        current_count += 1
        if start_frame is None:
            start_frame = df[frame_col][i]
    else:
        if start_frame is not None:  # فقط اگر start_frame مقداردهی شده باشد
            total_frame = df[frame_col][i] - start_frame + 1
            if total_frame >= 3:  # اگر تعداد فریمها بیشتر از 2 باشد (یعنی حداقل 3 فریم متوالی)
                results_list.append({
                    'Residue1': df[residue1_col][i],
                    'Residue3': df[residue3_col][i],
                    'hbond count': current_count + 1,
                    'first frame': start_frame,
                    'final frame': df[frame_col][i],
                    'total frame': total_frame
                })
        
        current_count = 0
        start_frame = None

# بررسی آخرین گروه متوالی
if start_frame is not None:  # فقط اگر start_frame مقداردهی شده باشد
    total_frame = df[frame_col].iloc[-1] - start_frame + 1
    if total_frame >= 3:
        results_list.append({
            'Residue1': df[residue1_col].iloc[-1],
            'Residue3': df[residue3_col].iloc[-1],
            'hbond count': current_count + 1,
            'first frame': start_frame,
            'final frame': df[frame_col].iloc[-1],
            'total frame': total_frame
        })

# تبدیل لیست به DataFrame
results = pd.DataFrame(results_list)

# ذخیره نتایج در یک فایل CSV جدید
results.to_csv(r'C:\Users\Lenovo\Desktop\biophys\5\run2\5H2O2_run2_repeated.csv', index=False)
```

```python
import pandas as pd

# بارگذاری فایل CSV
df = pd.read_csv(r'C:\Users\Lenovo\Desktop\biophys\5\run2\amino_hbond_5H2O2_run2.csv')

# تعریف ستونها
frame_col = 'Frame'
residue1_col = 'Residue1_Residue'
residue3_col = 'Residue3_Residue'

# ایجاد یک کلید ترکیبی برای مقایسه
df['Residue_Key'] = df[residue1_col].astype(str) + '_' + df[residue3_col].astype(str)

# گروهبندی بر اساس کلید ترکیبی و فریم
grouped = df.groupby('Residue_Key')[frame_col].agg(['min', 'max', 'count']).reset_index()

# فیلتر کردن گروههایی که حداقل 3 فریم دارند
results = grouped[grouped['count'] >= 1]

# تبدیل نتایج به DataFrame مناسب
final_results = pd.DataFrame({
    'hbond count': results['count'],
    'Residue1': results['Residue_Key'].str.split('_').str[0],
    'Residue3': results['Residue_Key'].str.split('_').str[1],
    'first frame': results['min'],
    'final frame': results['max'],
    'total frame': results['max'] - results['min'] + 1
})

# ذخیره نتایج در یک فایل CSV جدید
final_results.to_csv(r'C:\Users\Lenovo\Desktop\biophys\5\run2\1H2O2_run2_repeated2.csv', index=False)
```
