```python
import pandas as pd
import os
import matplotlib.pyplot as plt

# نام فایل CSV ورودی
input_file_path = r"C:\Users\Lenovo\Desktop\10H2O2\run1\full_preprocessed_10H2O2_run1.csv"  # نام فایل CSV خود را به درستی وارد کنید

# خواندن فایل CSV
df = pd.read_csv(input_file_path)

# اطمینان از وجود دایرکتوری برای ذخیره فایلهای خروجی
output_dir = 'output_files_run2'
os.makedirs(output_dir, exist_ok=True)

# گروهبندی بر اساس ستونهای Residue و Residue Number
grouped_df = df.groupby(['Residue', 'Residue Number'])

# ذخیره هر گروه در یک فایل CSV جداگانه و رسم نمودار
for (residue, residue_number), group in grouped_df:
    # ایجاد نام فایل با استفاده از Residue
    residue_name = residue.split('::')[1]  # جداسازی اسم Residue با استفاده از ::
    
    # نام فایل خروجی بر اساس Residue
    output_file_name = os.path.join(output_dir, f"Residue_{residue}_Number_{residue_number}.csv".replace('::', '-'))
    
    # ذخیره گروه در فایل CSV
    group.to_csv(output_file_name, index=False)
    
    # رسم نمودار TOTAL بر حسب Frame #
    plt.figure(figsize=(10, 6))
    plt.plot(group['Frame #'], group['TOTAL'], marker='o', linestyle='-', color='blue', label='TOTAL')
    plt.title(f'TOTAL vs Frame # for Residue {residue_name} (Number: {residue_number})')
    plt.xlabel('Frame #')
    plt.ylabel('TOTAL')
    plt.grid()
    plt.legend()
    plt.xticks(rotation=45)

    # ذخیره نمودار به صورت تصویر
    plot_file_name = os.path.join(output_dir, f"Residue_{residue}_Number_{residue_number}.png".replace('::', '-'))
    plt.savefig(plot_file_name)
    plt.close()  # بستن شکل بعد از ذخیره

print("فایلهای جداگانه و نمودارها با موفقیت ذخیره شدند.")
```
