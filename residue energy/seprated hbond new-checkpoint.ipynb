```python
# نام فایل ورودی و خروجی را مشخص کنید
input_file = r"C:\Users\Lenovo\Desktop\New folder\hbond1.ndx"
output_file =  r"C:\Users\Lenovo\Desktop\New folder\output_hbond1.ndx"

# فایل ورودی را باز کنید و خطوط آن را بخوانید
with open(input_file, 'r') as file:
    lines = file.readlines()

# لیستی برای نگه‌داشتن خطوط مورد نظر
filtered_lines = []

# هر خط را بررسی کنید
for line in lines:
    # اگر خط فقط شامل اعداد باشد، آن را به لیست اضافه کنید
    if line.strip().isdigit():
        filtered_lines.append(line)

# فایل خروجی را باز کنید و خطوط فیلتر شده را بنویسید
with open(output_file, 'w') as file:
    file.writelines(filtered_lines)

print("عملیات با موفقیت انجام شد.")
```

```python
# نام فایل ورودی و خروجی را مشخص کنید
input_file = r"C:\Users\Lenovo\Desktop\New folder\hbond2.ndx"
output_file =  r"C:\Users\Lenovo\Desktop\New folder\output_hbond2.ndx"

# باز کردن فایل ورودی و خواندن خطوط
with open(input_file, 'r') as file:
    lines = file.readlines()

# باز کردن فایل خروجی برای نوشتن
with open(output_file, 'w') as file:
    for line in lines:
        # اگر خط فقط شامل اعداد باشد، آن را بنویسید
        if not line.startswith('[') or 'hbonds_Protein-H2O2_frame_' not in line:
            file.write(line)
print("عملیات با موفقیت انجام شد.")
```
