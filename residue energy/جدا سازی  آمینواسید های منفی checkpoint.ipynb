```python
import pandas as pd
import os

# تعریف مسیر فایل CSV
file_path = 'full_preprocessed_5H2O2_run1.csv'  # مسیر صحیح فایل CSV خود را بگذارید

# خواندن دادهها از فایل CSV
df = pd.read_csv(file_path)

# جدا کردن residue هایی با TOTAL منفی
negative_data = df[df['TOTAL'] < 0]

# بررسی اینکه آیا دادههای منفی وجود دارد یا نه
if negative_data.empty:
    print("هیچ داده منفی وجود ندارد.")
else:
    # گروهبندی دادهها بر اساس Residue
    grouped_data = negative_data.groupby('Residue')

    
# ایجاد پوشه "separated" اگر نیاز است
    output_folder = 'separated'
    os.makedirs(output_folder, exist_ok=True)  # این خط دایرکتوری را ایجاد میکند

# ذخیره هر گروه در فایل CSV جداگانه
for Residue, group in grouped_data:
    # نام فایل برای هر residue
    file_name = f"{Residue}.csv".replace('::', '-')  # جایگزینی / با - برای نام گذاری فایل
    file_path = os.path.join(output_folder, file_name)  # استفاده از دایرکتوری ایجاد شده

    # ذخیره دادهها در فایل CSV
    group.to_csv(file_path, index=False)
```
