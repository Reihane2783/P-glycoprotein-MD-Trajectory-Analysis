```python
import pandas as pd

file_path =r"C:\Users\Lenovo\Desktop\delata total_hbond\FINAL_RESULTS_MMPBSA_run1.csv" 

with open(file_path, 'r') as file:
    lines = file.readlines()

# جستجوی هندسی برای پیدا کردن خط مربوط به "Ligand Energy Terms" و "Delta Energy Terms"
start_index = end_index = -1
for i, line in enumerate(lines):
    if "Ligand Energy Terms" in line:
        start_index = i + 1  # شروع از خط بعد از "Ligand Energy Terms"
    elif "Delta Energy Terms" in line:
        end_index = i  # بخشی که باید متوقف شود
        break

# اطمینان از اینکه هر دو اندیس معتبر هستند
if start_index != -1 and end_index != -1:
    # استخراج دادهها در بازه مشخص شده
    data = []
    for line in lines[start_index:end_index]:
        if line.strip():  # اگر خط خالی نیست
            values = line.split(',')
            if len(values) >= 2:  # بررسی وجود حداقل ۲ ستون
                frame = values[0].strip()  # استخراج Frame #
                total = values[-1].strip()  # استخراج Total
                data.append({'Frame #': frame, 'TOTAL': total})

    # تبدیل دادهها به DataFrame
    results_df = pd.DataFrame(data)

    # ذخیره نتایج در فایل CSV
    output_path = 'separated Ligand3.csv'  # مسیر ذخیره فایل خروجی خود را اینجا وارد کنید
    results_df.to_csv(output_path, index=False)

    print("Process completed successfully!")
else:
    print("Could not find the specified sections.")
```

```python
import pandas as pd

file_path =r"C:\Users\Lenovo\Desktop\delata total_hbond\FINAL_RESULTS_MMPBSA_run3.csv" 

with open(file_path, 'r') as file:
    lines = file.readlines()

# جستجوی خطی که "Delta Energy Terms" را شامل میشود
start_index = -1
for i, line in enumerate(lines):
    if "Delta Energy Terms" in line:
        start_index = i + 1  # شروع از خط بعد از "Delta Energy Terms"
        break

# اطمینان از اینکه اندیس معتبر است
if start_index != -1:
    # استخراج دادهها از خط بعد از Delta Energy Terms
    data = []
    for line in lines[start_index:]:
        if line.strip():  # اگر خط خالی نیست
            values = line.split(',')
            if len(values) >= 2:  # بررسی وجود حداقل ۲ ستون
                frame = values[0].strip()  # استخراج Frame #
                total = values[-1].strip()  # استخراج Total
                data.append({'Frame #': frame, 'Total': total})

    # تبدیل دادهها به DataFrame
    results_df = pd.DataFrame(data)

    # ذخیره نتایج در فایل CSV
    output_path = 'separated Delta run3.csv'  # مسیر ذخیره فایل خروجی خود را اینجا وارد کنید
    results_df.to_csv(output_path, index=False)

    print("Process completed successfully!")
else:
    print("Could not find the specified section.")
```
