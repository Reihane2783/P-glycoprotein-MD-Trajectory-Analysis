```python
import pandas as pd
import re

# بارگذاری فایل CSV
df = pd.read_csv(r'C:\Users\Lenovo\Desktop\total hbond\count_hbond\DISPcombined_output_final_1H2O2_run3.csv')

# تابعی برای تغییر فرمت داده‌ها
def rearrange_amino(amino):
    if 'H2O2' in amino:
       return amino
    # پیدا کردن حروف و اعداد
    letters = re.findall('[a-zA-Z]+', amino)
    numbers = re.findall('[0-9]+', amino)
    
    # ترکیب حروف و اعداد به ترتیب مورد نظر
    if letters and numbers:
        return letters[0] + ''.join(numbers)
    return amino  # اگر فرمت غیر منتظره بود، داده را بدون تغییر برگردانید

# اعمال تابع به ستون 'amino'
df['Amino 1'] = df['Amino 1'].apply(rearrange_amino)

# ذخیره تغییرات در یک فایل CSV جدید
df.to_csv(r'C:\Users\Lenovo\Desktop\total hbond\count_hbond\TOTAL FRAME 1H2O2_RUN3.csv', index=False)
```
