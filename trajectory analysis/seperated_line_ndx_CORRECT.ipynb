FOR FIND RESIDUE NAME

```python
import pandas as pd

# نام فایل متنی که داده‌ها در آن قرار دارند
input_file ='hbond_extra_mix_open_run1 - Copy.ndx'
# نام فایل CSV که داده‌ها به آن ذخیره می‌شود
output_file = 'hbond_extra_mix_open_run1 - Copy------.csv'

# خواندن داده‌ها از فایل متنی
data = []
with open(input_file, 'r') as file:
    for line in file:
        # حذف کاراکترهای اضافی و تقسیم داده‌ها بر اساس فضای خالی
        values = line.strip().split()
        #if len(values) == 3:  # اطمینان از اینکه سه ستون وجود دارد
        data.append(values)

# ساخت DataFrame از داده‌ها
df = pd.DataFrame(data, columns=['residue1', 'residue2','residue3'])
#df = df[(df['hbond']!=0).all]
# ذخیره داده‌ها در فایل CSV
df.to_csv(output_file, index=False)

print(f"داده‌ها با موفقیت در '{output_file}' ذخیره شد.")
```

```python
import pandas as pd

# نام فایل متنی که داده‌ها در آن قرار دارند
input_file ='hbond_extra_mix_res_run3.xvg'
# نام فایل CSV که داده‌ها به آن ذخیره می‌شود
output_file = 'hbond_extra_mix_open_run3.csv'

# خواندن داده‌ها از فایل متنی
data = []
with open(input_file, 'r') as file:
    for line in file:
        # حذف کاراکترهای اضافی و تقسیم داده‌ها بر اساس فضای خالی
        values = line.strip().split()
        if len(values) == 2:  # اطمینان از اینکه سه ستون وجود دارد
            data.append(values)

# ساخت DataFrame از داده‌ها
df = pd.DataFrame(data, columns=['frame', 'hbond'])
#df = df[df['hbond']!=0]
# ذخیره داده‌ها در فایل CSV
df.to_csv(output_file, index=False)

print(df)
```
