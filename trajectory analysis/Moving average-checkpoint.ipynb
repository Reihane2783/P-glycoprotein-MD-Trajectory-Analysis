```python
def filter(data_name):
    data_name =data_name[~data_name['Residue'].str.contains('L::H2O2', case=False, na=False)]
    data_name =data_name[~data_name['Residue'].str.contains('L::H', case=False, na=False)]
    data_name['Residue']=data_name['Residue'].str.replace('R::','',regex=False)
    data_name['Residue']=data_name['Residue'].str.replace(':','',regex=False)
    data_name['Residue']=data_name['Residue'].str.replace('::','',regex=False)
    return data_name
```

```python
def line_with_moving_average(data_name):
    filtered_data_name = filter(data_name)
    data_name_unique=filtered_data_name['Residue'].unique().tolist()
    for i in data_name_unique:
        filtering = filtered_data_name[(filtered_data_name['Residue']==i)].copy()
        filtering['Moving Average'] = filtering['TOTAL'].rolling(window=10).mean()
        plt.plot(filtering['Frame #'],filtering['TOTAL'],color='b',label='Accurate')
        plt.plot(filtering['Frame #'],filtering['Moving Average'],color='r',label='Moving Average')
        plt.xlabel('Frame')
        plt.ylabel('TOTAL')
        plt.legend()
        plt.title(f"{i}")
        plt.savefig(f"{i}" , dpi=500)
        plt.show()
```
