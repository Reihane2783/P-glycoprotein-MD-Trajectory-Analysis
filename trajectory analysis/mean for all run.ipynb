```python
folder_path = r'C:\Users\Lenovo\Desktop\output+all\output_directory_10H2O2_run3'

for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)
        df = pd.read_csv(file_path)
        df['run'] = 3
        df.to_csv(file_path, index=False)

print("ستون 'ران' به تمامی فایل‌های CSV اضافه شد.")
```

```python
# مسیر فایل‌های CSV را مشخص کنید
file_path = r'C:\Users\Lenovo\Desktop\output+all\all output/*.csv'
# لیست کردن همه فایل‌های CSV
all_files = glob.glob(file_path)

# لیست برای ذخیره DataFrame های خوانده شده
dataframes = []

# خواندن هر فایل و افزودن به لیست
for filename in all_files:
    df = pd.read_csv(filename)
    dataframes.append(df)

# ادغام همه DataFrame ها به یک DataFrame واحد
merged_df = pd.concat(dataframes, ignore_index=True)

# ذخیره DataFrame ادغام شده به یک فایل CSV جدید
merged_df.to_csv('merged_file.csv', index=False)
print("All files have been merged successfully!")
```
