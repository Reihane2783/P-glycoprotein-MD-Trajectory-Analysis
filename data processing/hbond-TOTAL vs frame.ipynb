```python
import pandas as pd
import matplotlib.pyplot as plt

file_path_1=r'C:\Users\Lenovo\Desktop\final\run3\full_preprocessed_run3.csv'
file_path_2 = r'C:\Users\Lenovo\Desktop\hbond\seperated.hbondrun3.csv'
data1 = pd.read_csv(file_path_1)
data2 = pd.read_csv(file_path_2)

frame_column_1=data1['Frame #']
total_column=data1['TOTAL']
frame_column_2 = data2['frame']
hbond_column = data2['hbond']


plt.figure(figsize=(10, 6))
plt.plot(frame_column_1, total_column, marker='o', linestyle='-', label='Total')
plt.plot(frame_column_2, hbond_column, marker='x', linestyle='--', label='Hbond')
plt.title('Delta Total and Hbond vs Frame 10 H2O2 run3')
plt.xlabel('Frame')
plt.ylabel('Values')
plt.legend()
plt.grid()
plt.savefig('Frame_hbond_Delta Total energy run3.png',dpi=500)
plt.show()
```
