```python
import pandas as pd
import os

path = r'C:\Users\Lenovo\Desktop\output+all\output_directory_10H2O2_run1'

   # لیست کردن تمام فایل‌های CSV در دایرکتوری
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]

for file in csv_files:
       # بارگذاری فایل CSV
    file_path = os.path.join(path, file)
    df = pd.read_csv(file_path)

       # بررسی وجود ستون 'moving ave'
    if 'Moving Average' in df.columns:
           # مرتب‌سازی بر اساس ستون 'moving ave'
        df_sorted = df.sort_values(by='Moving Average')

           # ذخیره‌سازی فایل مرتب شده
        sorted_file_path = os.path.join(path, f'sorted_allrun10_{file}')
        df_sorted.to_csv(sorted_file_path, index=False)

        print(f'فایل {file} مرتب شد و به {sorted_file_path} ذخیره شد.')
    else:
        print(f'ستون "moving ave" در فایل {file} وجود ندارد.')
```
