```python
# نام فایل ورودی و خروجی
input_file = r'C:\Users\Lenovo\Desktop\hbond.ndx'  # نام فایل ndx شما
output_file =r'C:\Users\Lenovo\Desktop\modified_data.ndx'  # نام فایل خروجی

# باز کردن فایل ورودی برای خواندن و فایل خروجی برای نوشتن
with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
    current_section = None

    # پردازش خط به خط
    for line in infile:
        line = line.strip()
        if line.startswith('[') and line.endswith(']'):
            # اگر خط یک بخش است، آن را ذخیره کنیم
            current_section = line[1:-1]  # حذف []
        elif line:  # اگر خط خالی نیست
            # اضافه کردن خط جدید با ترتیب معکوس
            parts = line.split()
            if len(parts) == 3:
                if current_section:
                    outfile.write(f'[{current_section}]\n')  # نوشتن نام بخش

                new_line = f", {parts[0]}  , {parts[1]} , {parts[2]} , "
                # نوشتن داده‌های معکوس و نام بخش به فایل خروجی
                outfile.write(new_line + '\n')

print("داده‌ها با موفقیت اصلاح شدند.")
print(output_file)
```

```python
import pandas as pd

# خواندن فایل اول (با ۳ ستون عددی)
file1 =  r'C:\Users\Lenovo\Desktop\hbondrun3.csv'
data1 = pd.read_csv(file1)

# خواندن فایل دوم (با نام آمینو اسید و عدد)
file2 =r'C:\Users\Lenovo\Desktop\HBOND\conf.csv'  # نام فایل دوم
data2 = pd.read_csv(file2)

# فرض می‌کنیم که ستون‌های فایل اول به نام 'num1', 'num2', 'num3' هستند
# و ستون‌های فایل دوم به نام 'amino_acid' و 'value' هستند

# ایجاد یک دیکشنری برای نگهداری تطابق عددها با نام آمینو اسیدها
amino_acid_dict = dict(zip(data2['number'], data2['Residue']))

# تابعی برای پیدا کردن نام آمینو اسید
def find_amino_acid(num):
    return amino_acid_dict.get(num)

# اعمال تابع برای هر عدد در هر ستون فایل اول

data1['amino1'] = data1['Residue1'].apply(find_amino_acid)
data1['amino2'] = data1['Residue2'].apply(find_amino_acid)
data1['amino3'] = data1['Residue3'].apply(find_amino_acid)

aminoframe_file =  r'C:\Users\Lenovo\Desktop\aminoatom_run3.csv'
data1.to_csv(aminoframe_file, index=False)

print("نتایج در فایل output.csv ذخیره شد.")
```

```python
import pandas as pd

# بارگذاری فایل CSV موجود
input_file = r"C:\Users\Lenovo\Desktop\aminoatom_run1.csv" # نام فایل ورودی
data = pd.read_csv(input_file)

# نام ستون‌های حاوی آمینو اسیدها
residue_columns = ['amino1','amino2','amino3']  # لیست ستون‌ها

# ایجاد یک DataFrame جدید برای ذخیره نتایج
result_df = pd.DataFrame()

# شمارش تکرارهای هر آمینو اسید در هر ستون
for column in residue_columns:
    # شمارش تعداد تکرارهای آمینو اسیدها در ستون
    counts = data[column].value_counts()
    
    # تبدیل به DataFrame و تغییر نام ستون‌ها
    counts_df = counts.reset_index()
    counts_df.columns = ['Amino Acid', column + ' Count']
    
    # ادغام نتایج با DataFrame اصلی
    if result_df.empty:
        result_df = counts_df
    else:
        result_df = pd.merge(result_df, counts_df, on='Amino Acid', how='outer')

# ذخیره‌سازی به یک فایل CSV جدید
output_file = r"C:\Users\Lenovo\Desktop\caunt_run1.csv"   # نام فایل خروجی
result_df.to_csv(output_file, index=False, encoding='utf-8-sig')

print(f'فایل {output_file} با موفقیت ایجاد شد.')
```
